### **Clinical Range Check for GCS**

GCS is clinically scored from `3` to `15`, so values below 3 or above 15 would be considered invalid or unrealistic for this column.

If values such as `2` or `16` appear, they should not be accepted as normal data. Instead, they should be flagged and handled carefully. For this notebook, I would convert out-of-range GCS values to `NaN` and then decide whether to impute them based on the pattern of missing data.

This approach is safer than deleting the entire patient row, because the rest of the patient’s information may still be useful. It is also more clinically justified than simply forcing values to the nearest boundary without knowing whether the entry was a true error.


```python
# I check for GCS values outside the valid clinical range.
# Since GCS should only be between 3 and 15, values below 3 or above 15
# are flagged as invalid before deciding how to clean them.

# Step 2: Flag physiologically impossible values
# GCS valid range: 3–15
invalid_gcs = df[(df['GCS'] < 3) | (df['GCS'] > 15)]
print(f"GCS values outside valid range (3–15): {len(invalid_gcs)} rows")
print(invalid_gcs[['ID', 'GCS']].head(10))
```

```python
# I replace invalid GCS values with NaN instead of deleting the entire row.
# This keeps the rest of the patient's data while marking the incorrect GCS entry as missing.

# Step 3: Replace invalid values with NaN
df.loc[(df['GCS'] < 3) | (df['GCS'] > 15), 'GCS'] = np.nan
print(f"NaN count after range filter: {df['GCS'].isnull().sum()}")
```

## 5. Imputation — Filling Missing Values

### **Choosing an Imputation Strategy**

After invalid GCS values were converted to `NaN`, I needed to decide how to handle the missing values. Common imputation options include the mean, median, and mode.

Three common imputation strategies:

| Strategy | Use when | Risk |
|----------|----------|------|
| **Mean** | Data is normally distributed, no extreme outliers | Sensitive to outliers — pulls the mean |
| **Median** | Data is skewed or has outliers | More robust, but loses distributional info |
| **Mode** | Categorical or count data | Only works if one value dominates |

For this dataset, the median is the most appropriate choice for GCS because GCS is a bounded clinical score from `3` to `15`, and ED datasets are often skewed toward higher scores such as `15`. The **median** is less affected by extreme or unusual values than the mean, so it is a safer option for this type of clinical variable.

I avoided deleting entire rows because the rest of the patient information may still be useful for later analysis.


```python
# Compute the median and use it to fill NaN values
gcs_median = df['GCS'].median()
print(f"GCS median (before imputation): {gcs_median}")

df['GCS'] = df['GCS'].fillna(gcs_median)

print(f"GCS after imputation:")
print(df['GCS'].describe())
print(f"Remaining NaNs: {df['GCS'].isnull().sum()}")
```

## 7. Temperature — A Special Case

### Temperature Cleaning Example

The `Temp` column demonstrates a common clinical data issue: unit inconsistency. Some values may be entered as plain Celsius values, some may include a `C` suffix, and some may be recorded in Fahrenheit. This example shows why clinical data cleaning must consider both formatting and units before analysis.

```python
# First, look at the raw values
print("Temp unique values (sample):")
print(df['Temp'].unique()[:20])
```

```python
# Write a function to standardise everything to Celsius
def to_celsius(val):
    """Convert temperature value to Celsius float.
    Handles: plain numbers, strings ending in 'C', strings ending in 'F'.
    Returns NaN for anything unrecognisable.
    """
    if pd.isnull(val):
        return np.nan
    val_str = str(val).strip()
    try:
        if val_str.endswith('C'):
            return float(val_str[:-1])          # remove 'C', convert
        elif val_str.endswith('F'):
            return (float(val_str[:-1]) - 32) * 5 / 9  # Fahrenheit to Celsius
        else:
            return float(val_str)               # assume Celsius already
    except ValueError:
        return np.nan                           # anything else -> NaN

df['Temp'] = df['Temp'].apply(to_celsius)

print("After standardisation:")
print(df['Temp'].describe())
```

```python
# Apply range filter: valid body temperature 32–43°C
df.loc[(df['Temp'] < 32) | (df['Temp'] > 43), 'Temp'] = np.nan

# Impute with median
temp_median = round(df['Temp'].median(), 1)
df['Temp'] = df['Temp'].fillna(temp_median)

print(f"Temp after cleaning: min={df['Temp'].min():.1f}, max={df['Temp'].max():.1f}, NaNs={df['Temp'].isnull().sum()}")
```

### **Preparing SBP for MAP Cleaning**


Since our assigned column is `MAP`, we first prepared `SBP` because MAP is calculated using both systolic and diastolic blood pressure. `SBP` stands for systolic blood pressure, which is the upper number in a blood pressure reading.

The valid SBP range used for this task is `50-250 mmHg`. Values outside this range were treated as invalid and converted to `NaN`.

After invalid SBP values were identified, we used **median imputation** to fill the missing SBP entries. The median was selected because blood pressure values can contain outliers, and the median is less affected by extreme values than the mean. This allowed the SBP column to remain usable for the later MAP calculation.

```python
# Preparing SBP for MAP calculation
# SBP = Systolic Blood Pressure, the upper number in a blood pressure reading.
# Valid SBP range for this task: 50-250 mmHg.

print("=== Preparing SBP for MAP calculation ===")

# Inspecting SBP before cleaning
print("SBP dtype before cleaning:", df["SBP"].dtype)
print("SBP unique values sample:", df["SBP"].unique()[:20])
print("SBP missing values before cleaning:", df["SBP"].isnull().sum())

# Convert SBP to numeric
df["SBP"] = pd.to_numeric(df["SBP"], errors="coerce")

print("\nSBP dtype after conversion:", df["SBP"].dtype)
print("SBP missing values after conversion:", df["SBP"].isnull().sum())

# Flag invalid SBP values
invalid_sbp = df[(df["SBP"] < 50) | (df["SBP"] > 250)]

print(f"\nSBP values outside valid range (50-250): {len(invalid_sbp)} rows")
print(invalid_sbp[["ID", "SBP"]].head(10))

# Replace invalid SBP values with NaN
df.loc[(df["SBP"] < 50) | (df["SBP"] > 250), "SBP"] = np.nan

print("\nSBP missing values after range filtering:", df["SBP"].isnull().sum())

# Impute missing SBP values using the median only if needed
# Median is used as the fallback because blood pressure values can contain outliers.
sbp_missing_before_imputation = df["SBP"].isnull().sum()

if sbp_missing_before_imputation > 0:
    sbp_median = df["SBP"].median()
    df["SBP"] = df["SBP"].fillna(sbp_median)

    print(f"\nMedian SBP used for imputation: {sbp_median}")
    print("SBP missing values after median imputation:", df["SBP"].isnull().sum())
else:
    print("\nNo SBP values required median imputation.")

print("\nSBP summary after preparation:")
print(df["SBP"].describe())
```

### **Preparing DBP for MAP Cleaning**

We also prepared `DBP` because MAP is calculated using both systolic and diastolic blood pressure. `DBP` stands for diastolic blood pressure, which is the lower number in a blood pressure reading.

The valid DBP range used for this task is `30-150 mmHg`. Values outside this range were treated as invalid and converted to `NaN`.

After invalid and missing DBP values were identified, we used **median imputation** to fill the missing DBP entries. The median was selected because blood pressure values can contain outliers, and the median is less affected by extreme values than the mean. This allowed the DBP column to remain usable for the later MAP calculation.

```python
# Preparing DBP for MAP calculation
# DBP = Diastolic Blood Pressure, the lower number in a blood pressure reading.
# Valid DBP range for this task: 30-150 mmHg.

print("=== Preparing DBP for MAP calculation ===")

# Inspecting DBP before cleaning
print("DBP dtype before cleaning:", df["DBP"].dtype)
print("DBP unique values sample:", df["DBP"].unique()[:20])
print("DBP missing values before cleaning:", df["DBP"].isnull().sum())

# Convert DBP to numeric
df["DBP"] = pd.to_numeric(df["DBP"], errors="coerce")

print("\nDBP dtype after conversion:", df["DBP"].dtype)
print("DBP missing values after conversion:", df["DBP"].isnull().sum())

# Flag invalid DBP values
invalid_dbp = df[(df["DBP"] < 30) | (df["DBP"] > 150)]

print(f"\nDBP values outside valid range (30-150): {len(invalid_dbp)} rows")
print(invalid_dbp[["ID", "DBP"]].head(10))

# Replace invalid DBP values with NaN
df.loc[(df["DBP"] < 30) | (df["DBP"] > 150), "DBP"] = np.nan

print("\nDBP missing values after range filtering:", df["DBP"].isnull().sum())

# Impute missing DBP values using the median
# Median is used because blood pressure values can contain outliers,
# and the median is less affected by extreme values than the mean.
dbp_median = df["DBP"].median()

df["DBP"] = df["DBP"].fillna(dbp_median)

print(f"\nMedian DBP used for imputation: {dbp_median}")
print("DBP missing values after median imputation:", df["DBP"].isnull().sum())

print("\nDBP summary after imputation:")
print(df["DBP"].describe())

```

### Cleaning MAP

Our assigned column is `MAP`, which stands for mean arterial pressure. MAP represents the average pressure in a patient’s arteries during one cardiac cycle and is useful for assessing overall blood perfusion.

For this task, the expected MAP range is `40-180 mmHg`. However, MAP is a derived clinical value, so we did not treat it as a completely independent column. Instead, we recalculated MAP using the cleaned `SBP` and `DBP` columns with the formula:

`MAP = (SBP + 2 × DBP) / 3`

This formula-based recalculation was used as the primary cleaning method. It is more clinically meaningful than immediately applying mean, median, or mode imputation because it uses the relationship between the blood pressure variables.

After recalculation, we checked for missing and out-of-range MAP values. If any MAP values remained missing after recalculation, median imputation would be used as a fallback because the median is less affected by extreme values than the mean. In this run, no MAP values were missing after recalculation, so direct MAP median imputation was not required.

Any MAP values outside the expected range were reviewed alongside their cleaned SBP and DBP values. If the supporting blood pressure values were valid, the MAP value was preserved and flagged as clinically significant rather than automatically treated as a data error.

```python
# Cleaning MAP
# MAP = Mean Arterial Pressure
# Valid MAP range for this task: 40-180 mmHg.
# MAP formula: MAP = (SBP + 2*DBP) / 3

print("=== Cleaning MAP ===")

# Inspect MAP before recalculation
print("MAP dtype before cleaning:", df["MAP"].dtype)
print("MAP unique values sample:", df["MAP"].unique()[:20])
print("MAP missing values before cleaning:", df["MAP"].isnull().sum())

# Convert MAP to numeric for inspection/documentation
df["MAP"] = pd.to_numeric(df["MAP"], errors="coerce")

print("\nMAP dtype after conversion:", df["MAP"].dtype)
print("MAP missing values after conversion:", df["MAP"].isnull().sum())

print("\nOriginal MAP summary before recalculation:")
print(df["MAP"].describe())

# Recalculate MAP from cleaned SBP and DBP
# This is the primary cleaning method because MAP is derived from SBP and DBP.
df["MAP"] = ((df["SBP"] + 2 * df["DBP"]) / 3).round(2)

print("\nMAP after recalculation from cleaned SBP and DBP:")
print(df["MAP"].describe())

# Check for MAP values outside the expected valid range
out_of_range_map = df[(df["MAP"] < 40) | (df["MAP"] > 180)]

print(f"\nMAP values outside 40-180 mmHg: {len(out_of_range_map)}")

if not out_of_range_map.empty:
    print(out_of_range_map[["ID", "SBP", "DBP", "MAP"]])
    print()
    print("These MAP values were reviewed using cleaned SBP and DBP.")
    print("If SBP and DBP are within valid ranges, the MAP value may represent")
    print("a critically ill patient rather than a data error, so it was preserved.")
```

```python
# Check if any MAP values are still missing after recalculation
map_missing_after_recalculation = df["MAP"].isnull().sum()

print("\nMissing MAP values after recalculation:", map_missing_after_recalculation)

# Use median imputation only as a fallback if MAP values remain missing
if map_missing_after_recalculation > 0:
    map_median = df["MAP"].median()
    df["MAP"] = df["MAP"].fillna(map_median)

    print(f"Fallback median MAP used for imputation: {map_median:.2f} mmHg")
    print("Missing MAP values after fallback median imputation:", df["MAP"].isnull().sum())
else:
    print("No MAP values were missing after recalculation, so MAP median imputation was not required.")

# Add a clinical flag for MAP values outside the expected range
df["MAP_clinical_flag"] = "Within expected range"

df.loc[df["MAP"] < 40, "MAP_clinical_flag"] = "Low MAP - possible critical hypotension"
df.loc[df["MAP"] > 180, "MAP_clinical_flag"] = "High MAP - review clinically"

print("\nMAP clinical flag counts:")
print(df["MAP_clinical_flag"].value_counts())

print("\nFlagged MAP cases:")
print(df[df["MAP_clinical_flag"] != "Within expected range"][["ID", "SBP", "DBP", "MAP", "MAP_clinical_flag"]])

print("\nFinal MAP summary:")
print(df["MAP"].describe())
```

### **MAP Cleaning Result**

After cleaning the supporting blood pressure columns, we recalculated `MAP` using the formula:

`MAP = (SBP + 2 × DBP) / 3`

This produced no missing MAP values. One MAP value fell slightly below the expected range of `40-180 mmHg`, with a value of `39.33 mmHg`. However, the corresponding `SBP` and `DBP` values were both within their valid ranges, so this value was preserved as a clinically significant finding rather than treated as a data error.

Because no MAP values were missing after recalculation, direct median imputation for MAP was not required. Median imputation was only kept as a backup approach in case MAP could not be recalculated from SBP and DBP.
