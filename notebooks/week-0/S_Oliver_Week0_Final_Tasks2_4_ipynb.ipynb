## Task 3: Clinical Context Write-Up

Chosen vital sign: Mean Arterial Pressure (MAP)

Mean arterial pressure (MAP) is the average pressure within the arteries throughout one cardiac cycle and serves as a key indicator of whether adequate blood flow is being maintained to vital organs such as the brain, heart and kidneys. MAP is calculated using the formula MAP = (SBP + 2 × DBP) / 3, where SBP is systolic blood pressure and DBP is diastolic blood pressure. This equation is weighted toward diastolic pressure because the heart spends more time in diastole, the resting and filling phase, than in systole, the contracting phase. A normal MAP typically falls around 70-100 mmHg, while a MAP below 65 mmHg is often considered abnormal in critically ill patients and may indicate hypotension, poor organ perfusion or shock risk, warranting urgent clinical review. A MAP above approximately 110 mmHg may also be abnormal and can suggest increased vascular pressure, severe hypertension or significant cardiovascular strain. However, MAP should always be considered alongside other clinical indicators, including heart rate, oxygen saturation, respiratory rate, neurological status and the patient’s overall condition. In an emergency department setting, an abnormal MAP is a clinically significant finding that should prompt thorough patient assessment rather than be dismissed as a recording error.


## Task 4: Unincluded Vital Sign Write-Up

Chosen vital sign: Oxygen Saturation (SpO₂)

Oxygen saturation (SpO₂) measures the percentage of haemoglobin in the blood that is carrying oxygen. It is usually checked with a pulse oximeter placed on the finger, toe or earlobe. SpO₂ should be included in triage because it gives a quick idea of how well a patient is oxygenating and can help identify breathing problems early. For most healthy people, a normal SpO₂ is about 95-100%. In acutely ill adults who are not at risk of hypercapnic respiratory failure, the usual target range is around 94-98%. However, patients with conditions such as COPD may have a lower target range of 88-92%, because giving too much oxygen can worsen carbon dioxide retention. Low SpO₂ may indicate hypoxaemia and can be linked to conditions such as asthma attacks, pneumonia, pulmonary embolism, heart failure or shock. In triage, this is important because a patient with low oxygen levels may need urgent assessment, oxygen therapy, respiratory support, or escalation of care. At the same time, a reading of 100% should not always be taken as fully reassuring, especially if the patient is already receiving oxygen. Some patients can also have silent hypoxia, where their oxygen level is dangerously low even though they do not look severely distressed. For this reason, SpO₂ should always be interpreted alongside respiratory rate, work of breathing, heart rate, mental status and the patient’s overall clinical condition.

## Task 5: Digital Triage Pseudocode

For Task 5, I designed rule-based pseudocode for a digital emergency department triage support model. The model uses vital signs to calculate a risk score, apply critical override rules, flag abnormal values, and support nurse review.

```
START

1. Receive Patient Data

Input patient information:
    Age
    Gender
    GCS
    SBP
    DBP
    MAP
    Pulse
    Temperature
    Respiratory Rate
    FiO2
    SpO2, if available
    Pain score, if available
    Presenting complaint, if available

Initialize:
    risk_score = 0
    risk_level = "Unassigned"
    clinical_flags = empty list
    missing_data_flags = empty list
    critical_override = False


2. Check for Missing or Invalid Data

FOR each required vital sign:
    IF value is missing:
        Add the vital sign name to missing_data_flags
        Add "Incomplete data: manual nurse review needed" to clinical_flags

    IF value is impossible or non-numeric:
        Mark value as invalid
        Add "Invalid value detected" to clinical_flags

IF important data is missing:
    Continue processing available data
    Do not delete the patient record
    Recommend manual nurse review


3. Prepare Blood Pressure Values

IF MAP is missing AND SBP and DBP are available:
    Calculate MAP using:
        MAP = (SBP + 2 × DBP) / 3

IF MAP is abnormal but physiologically possible:
    Preserve the value
    Add "Abnormal MAP: clinical review needed" to clinical_flags


4. Check Consciousness Level Using GCS

IF GCS <= 8:
    Add 3 points to risk_score
    Add "Severely reduced consciousness" to clinical_flags
    Set critical_override = True

ELSE IF GCS is between 9 and 12:
    Add 2 points to risk_score
    Add "Moderately reduced consciousness" to clinical_flags

ELSE IF GCS is between 13 and 14:
    Add 1 point to risk_score
    Add "Mildly reduced consciousness" to clinical_flags

ELSE IF GCS = 15:
    Add 0 points
    Add "Normal consciousness level" if needed


5. Check Blood Pressure and Perfusion

IF MAP < 65:
    Add 3 points to risk_score
    Add "Low MAP / possible poor organ perfusion" to clinical_flags
    Set critical_override = True

IF SBP < 90:
    Add 3 points to risk_score
    Add "Low SBP / possible hypotension or shock risk" to clinical_flags
    Set critical_override = True

IF SBP > 180:
    Add 2 points to risk_score
    Add "Very high SBP / possible severe hypertension" to clinical_flags


6. Check Pulse

IF pulse < 50:
    Add 2 points to risk_score
    Add "Low pulse / possible bradycardia" to clinical_flags

ELSE IF pulse > 130:
    Add 2 points to risk_score
    Add "Very high pulse / urgent review needed" to clinical_flags

ELSE IF pulse > 100:
    Add 1 point to risk_score
    Add "Elevated pulse / possible tachycardia" to clinical_flags


7. Check Respiratory Rate

IF respiratory_rate < 8:
    Add 3 points to risk_score
    Add "Dangerously low respiratory rate" to clinical_flags
    Set critical_override = True

ELSE IF respiratory_rate > 30:
    Add 3 points to risk_score
    Add "Very high respiratory rate / respiratory distress risk" to clinical_flags
    Set critical_override = True

ELSE IF respiratory_rate > 20:
    Add 1 point to risk_score
    Add "Elevated respiratory rate" to clinical_flags


8. Check Temperature

IF temperature < 35:
    Add 2 points to risk_score
    Add "Hypothermia" to clinical_flags

ELSE IF temperature >= 39:
    Add 2 points to risk_score
    Add "High fever" to clinical_flags

ELSE IF temperature >= 38:
    Add 1 point to risk_score
    Add "Fever" to clinical_flags


9. Check Oxygenation and Oxygen Support

IF SpO2 is available:
    IF SpO2 < 90:
        Add 3 points to risk_score
        Add "Severe hypoxaemia" to clinical_flags
        Set critical_override = True

    ELSE IF SpO2 is between 90 and 94:
        Add 2 points to risk_score
        Add "Low oxygen saturation" to clinical_flags

IF FiO2 > 21:
    Add 1 point to risk_score
    Add "Patient may require oxygen support" to clinical_flags


10. Check for Multi-Sign Clinical Risk Patterns

IF MAP < 65 AND pulse > 100:
    Add 2 additional points to risk_score
    Add "Possible compensation or shock pattern: low MAP with high pulse" to clinical_flags
    Set critical_override = True

IF respiratory_rate > 30 AND SpO2 < 94:
    Add 2 additional points to risk_score
    Add "Possible respiratory compromise pattern" to clinical_flags
    Set critical_override = True

IF fever is present AND pulse > 100:
    Add 1 additional point to risk_score
    Add "Possible infection or systemic stress pattern" to clinical_flags


11. Assign Risk Level

IF critical_override = True:
    risk_level = "High Risk"

ELSE IF risk_score >= 6:
    risk_level = "High Risk"

ELSE IF risk_score is between 3 and 5:
    risk_level = "Moderate Risk"

ELSE:
    risk_level = "Low Risk"


12. Generate Explainable Triage Output

Display:
    Patient risk level
    Total risk score
    Clinical flags
    Missing data flags
    Reason for risk category
    Recommended triage action

IF risk_level = "High Risk":
    Recommend immediate nurse or clinician review

ELSE IF risk_level = "Moderate Risk":
    Recommend timely assessment and continued monitoring

ELSE IF risk_level = "Low Risk":
    Recommend routine triage assessment


13. Human-in-the-Loop Safety Check

Send output to triage nurse.

Triage nurse reviews:
    Vital signs
    Clinical flags
    Presenting complaint
    Patient appearance
    Clinical judgement

Final triage decision must be confirmed by the triage nurse.

END

```
